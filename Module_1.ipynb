# 3. Python Operators and Input/Output

Operators are symbols that perform operations on variables and values in Python. Python also provides the `input()` function to accept user input and the `print()` function to display output.

```python
# Arithmetic Operators

a = 25
b = 5

print("Addition:- ", a + b)
print("Subtraction:- ", a - b)
print("Multiplication:- ", a * b)
print("Division:- ", a / b)
print("Floor Division:", a // b)
print("Power:", a ** b)
```

```python
#user input and output

name = input("Enter name: ")
age = int(input("Enter age: "))

print("Hey!! ", name)
print("Your age is ", age)
```

## 4. Conditional Statements

Conditional statements allow Python programs to make decisions based on specific conditions. Python uses `if`, `elif`, and `else` statements to execute different blocks of code depending on whether a condition is met.

```python
marks = 64

if marks >= 80:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
elif marks >= 40:
    print("Grade C")
else:
    print("Fail")
```

## 5. Loops in Python

Loops are used to execute a block of code repeatedly. Python provides two main types of loops: the `for` loop and the `while` loop.

```python

#Calculate the sum of numbers

total = 0

for i in range(1, 10):
    total = total + i

print("Total:", total)
```

```python
#The while Loop

i = 1

while i <= 10:
    print(i)
    i = i + 1
```

```python
#Real-World Example: Sales Analysis

marks = [75, 85, 90, 65, 80]

total_marks = 0

for mark in marks:
    total_marks = total_marks + mark

average_marks = total_marks / len(marks)

print("Total Marks:", total_marks)
print("Average Marks:", average_marks)
```

```python
#Real-World Example: Sales Analysis

sales = [1200, 2500, 1800, 3200, 1500]

total_sales = 0

for amount in sales:
    total_sales = total_sales + amount

average_sales = total_sales / len(sales)

print("Total Sales:", total_sales)
print("Average Sales:", average_sales)
```

## 6. Lists and Dictionaries

Lists are used to store multiple items in a single variable, while dictionaries organize data into key-value pairs for easy access and management.

```python
# Creating a list
colors = ["Red", "Blue", "Green", "Yellow"]

print(colors)
print(type(colors))
```

```python

# Creating a list of student names
students = ["Rahul", "Priya", "Amit", "Sneha"]

# Accessing elements from the list
print(students[1])
print(students[3])
print(students[0])
print(students[-2])
```

```python

# Creating a list of subjects
subjects = ["Maths", "Science", "English"]

# Add an item
subjects.append("Computer")
print("After adding:", subjects)

# Update an item
subjects[0] = "History"
print("After updating:", subjects)

# Remove an item
subjects.remove("English")
print("After removing:", subjects)
```

```python

marks = [78, 92, 65, 88, 77]

print("Total Marks:", sum(marks))
print("Highest Marks:", max(marks))
print("Lowest Marks:", min(marks))
print("Number of Students:", len(marks))
print("Average Marks:", sum(marks) / len(marks))
```

```python

# Creating a dictionary of employee details
employee = {
    "name": "Priya",
    "age": 25,
    "salary": 35000
}

# Add a new key-value pair
employee["department"] = "IT"

# Update a value
employee["salary"] = 40000

# Remove a key-value pair
employee.pop("age")

print(employee)
```

```python
# Real-World Example: Employee Salaries

employees = [
    {"name": "Rahul", "salary": 25000},
    {"name": "Priya", "salary": 35000},
    {"name": "Amit", "salary": 30000}
]

total_salary = 0

for employee in employees:
    print(employee["name"], "-", employee["salary"])
    total_salary += employee["salary"]

print("Total Salary:", total_salary)
```

## 7. Functions in Python

A function is a reusable block of code designed to perform a particular task. Functions make programs more organized, reduce code repetition, and simplify maintenance.

```python
# Function with Return Value
def multiply(a, b):
    return a * b

result = multiply(5, 6)

print("Product:", result)
```

```python
# Real-World Example: Student Marks Analysis

def calculate_total(marks):
    return sum(marks)

def calculate_average(marks):
    return sum(marks) / len(marks)

marks = [75, 85, 90, 65, 80]

print("Total Marks:", calculate_total(marks))
print("Average Marks:", calculate_average(marks))
```

## 8. File Handling in Python

File handling in Python allows users to create, open, read, write, and manage files. The commonly used file modes are:

- `r`: Opens a file for reading.
- `w`: Opens a file for writing, replacing existing content or creating a new file.
- `a`: Opens a file to append new content to the end.

```python
# Writing employee details to a file
with open("employees.txt", "w") as file:
    file.write("Rahul - Salary: 25000\n")
    file.write("Priya - Salary: 35000\n")

print("Employee details saved successfully!")
```

```python
# Reading a File
with open("employees.txt", "r") as file:
    content = file.read()

print(content)
```

```python
# Appending Product Data to a File

with open("products.txt", "a") as file:
    file.write("Keyboard - Price: 1500\n")

print("New product details added!")
```

# 9. Data Science: Key Concepts and Practical Applications

## What is Data Science?

Data Science is a field that combines data, statistics, programming, and analytical methods to uncover meaningful insights and support decision-making.

Python is widely used in Data Science for cleaning and analyzing data, creating visualizations, and developing machine learning models.

## Real-World Applications of Data Science

1. Healthcare: Identifying potential diseases and analyzing patient health records.
2. Banking: Detecting fraudulent activities and monitoring financial transactions.
3. E-commerce: Suggesting relevant products based on customer preferences and purchase history.
4. Retail: Evaluating sales trends and understanding customer behavior.
5. Transportation: Forecasting traffic conditions and finding efficient routes.
6. Education: Assessing student performance and identifying learning patterns.

## Data Science Process

1. Data Collection: Gathering relevant data from various sources.
2. Data Cleaning: Identifying and correcting errors, missing values, and inconsistencies in the data.
3. Data Analysis: Examining data to identify patterns, trends, and useful insights.
4. Data Visualization: Presenting data through charts, graphs, and other visual representations.
5. Model Building: Developing statistical or machine learning models to make predictions.
6. Decision Making: Using data-driven insights to make informed decisions.

```python
# Student Performance Analysis

marks = [78, 92, 65, 88, 75]

total_marks = sum(marks)
average_marks = total_marks / len(marks)
highest_marks = max(marks)
lowest_marks = min(marks)

print("Student Marks:", marks)
print("Total Marks:", total_marks)
print("Average Marks:", average_marks)
print("Highest Marks:", highest_marks)
print("Lowest Marks:", lowest_marks)
```

- Data Collection: Gathered sales figures for analysis.
- Data Analysis: Calculated the total, average, and highest sales values.
- Decision-Making: Identified the highest sales value to support business decisions.

This example illustrates how Python helps perform basic data analysis and supports informed business decision-making.
